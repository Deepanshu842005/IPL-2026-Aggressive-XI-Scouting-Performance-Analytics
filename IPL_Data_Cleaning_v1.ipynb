```python
import pandas as pd 
df = pd.read_csv(r"C:\Users\deepa\Downloads\archive (2)\ipl_full_info(2008-2024).csv")
df.head()
```

```python
# 1. Recency Filter: Only 2022 to 2024
active_df = df[df['year'] >= 2022].copy()

# 2. Define Match Phases
def get_phase(over):
    if over < 6: return 'Powerplay'
    elif over < 15: return 'Middle Overs'
    else: return 'Death Overs'

active_df['match_phase'] = active_df['over'].apply(get_phase)

# 3. Categorize Bowlers (Add more names if needed)
spinners = ['Rashid Khan', 'YS Chahal', 'Kuldeep Yadav', 'R Ashwin', 'SP Narine', 'RA Jadeja', 'Varun Chakaravarthy', 'Ravi Bishnoi', 'Noor Ahmad']
active_df['bowler_type'] = active_df['Bowler'].apply(lambda x: 'Spinner' if x in spinners else 'Pacer')

print("✅ Feature Engineering Complete. Active players only!")
```

```python
import pandas as pd

# 1. Use your working path here
df = pd.read_csv(r"C:\Users\deepa\Downloads\archive (2)\ipl_full_info(2008-2024).csv") # <--- YOUR PATH HERE


# 2. Filter for Recency (2022-2024)
master_df = df[df['year'] >= 2022].copy()

# 3. Fix the runs calculation
master_df['runs_on_ball'] = master_df['batsman_run'] + master_df['extras']

# 4. Create Match Phases
master_df['match_phase'] = master_df['over'].apply(lambda x: 'Powerplay' if x < 6 else ('Middle Overs' if x < 15 else 'Death Overs'))

# 5. Define Bowler Wickets
wicket_types = ['caught', 'bowled', 'lbw', 'stumped', 'caught and bowled', 'hit wicket']
master_df['is_bowler_wicket'] = master_df['wicket_type'].apply(lambda x: 1 if x in wicket_types else 0)

# 6. SAVE the "Clean" version to your computer
master_df.to_csv('ipl_2026_scouting_cleaned.csv', index=False)

print("✅ DONE! You have turned 'Raw Ingredients' into a 'Clean Meal'.")
print("Check your folder for 'ipl_2026_scouting_cleaned.csv'")
```
